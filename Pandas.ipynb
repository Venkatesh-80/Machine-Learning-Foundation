**Group by in pandas**


```python
import numpy as np
import pandas as pd
```

```python
data={
    "Name": ["Aman","Saurav","Karan","Shubham","Abhishek",None,"Sameer"],
    "Age": [22,23,None,22,25,24,26],
    "Marks": [85,None,88,90,56,72,None],
    "Subject": ["Math","Science","Math","Science","Math","Science","Science"]
}
df=pd.DataFrame(data)
print(df)
```

```python
df.groupby("Subject")["Marks"].mean()   #By default the operation ignores NAN VAlues
```

```python
df.groupby("Subject")["Marks"].median()
```

```python
df.groupby("Subject")["Marks"].sum()
```

```python
df.groupby("Subject")["Name"].count()
```

```python
 #Multiple Aggregation
df.groupby("Subject")["Marks"].agg(["mean","median","sum"])
```

```python
df.groupby("Subject")["Marks"].agg(["min","max","count"])
```

```python
#Group Multiple Columns
df.groupby(["Subject","Name"])["Marks"].mean()
```

```python
data={
    "Name": ["Aman","Saurav","Karan","Shubham","Abhishek",None,"Sameer",None,"Venkatesh","Hari"],
    "Department": ["IT","CSE","Sales","HR",None,"CSE","Sales","HR","IT","CSE"],
    "Role": ["Manager","Developer","Analyst",None,"Data Scientist","Manager","Senior Manager","Manager",None,None],
    "Experience":[10,20,30,40,25,23,8,2,4,5],
    "Salary":[100000,23333,89999,99898,808097,90877,170000,20000,90000,100000],
    "Score": [10,20,30,40,50,60,70,80,90,99]
}
df=pd.DataFrame(data)
print(df)
```

```python
df.groupby("Department")["Salary"].mean()
```

```python
df.groupby("Role")["Score"].mean()
```

```python
df.groupby("Department")["Score"].count()
```

```python
df.groupby("Department")["Name"].count()
```

```python
df.groupby("Department").size()
```

```python
df.groupby("Department")["Salary"].agg(["mean","max","min","count"]) #Finding agg functions for salary based on Department
```

```python
df.groupby("Role")["Score"].agg(["mean","max","min","count"]) #Finding agg functions for score based on Role
```

```python
df.groupby(["Department","Role"])["Salary"].mean()
```

```python
df.groupby(["Department","Experience"])["Score"].mean()
```
